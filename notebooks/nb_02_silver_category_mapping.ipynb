```python
from pyspark.sql.functions import explode, split, col

cats = spark.table("silver_food_category")
mapping_t = spark.table("silver_category_mapping")
alt = spark.table("silver_plant_alternatives")

# 1) every environmental category name in the CSV is covered by our category table
owid_in_csv = (spark.read.csv("Files/food-environmental-impact/ghg-per-kg-poore.csv", header=True)
               .select(col("Entity").alias("owid_entity")).distinct())
in_category_table = cats.select(explode(split("owid_entities", ";")).alias("owid_entity"))

print("In CSV but missing from category table:", owid_in_csv.subtract(in_category_table).count())
print("In category table but not in CSV:", in_category_table.subtract(owid_in_csv).count())

# 2) every mapping row points to a real category
print("Mapping rows with unknown category:",
      mapping_t.join(cats, "food_category", "left_anti").count())

# 3) every alternative pair uses real categories on both sides
print("Alternative pairs with unknown animal category:",
      alt.join(cats.select(col("food_category").alias("animal_category")), "animal_category", "left_anti").count())
print("Alternative pairs with unknown plant category:",
      alt.join(cats.select(col("food_category").alias("plant_category")), "plant_category", "left_anti").count())
```
